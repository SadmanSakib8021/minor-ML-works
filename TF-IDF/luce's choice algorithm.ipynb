### Definition and Key Points

Luce's Axiom, also known as the Independence from Irrelevant Alternatives (IIA) axiom, is a fundamental concept in decision theory and choice modeling. The main points of the axiom are:

1. It states that the probability of selecting one item over another from a pool of many items is not affected by the presence or absence of other items in the pool.

2. This concept is described as "independence from irrelevant alternatives" (IIA).

3. Mathematically, it can be expressed as:

   P(i) = w_i / Σ_j w_j

   Where P(i) is the probability of selecting item i, w_i is the weight (measure of some typically salient property) of item i, and Σ_j w_j is the sum of weights of all j items.

### Mathematical Formulation

Let's break down the mathematical formulation of Luce's Axiom:

1. Probability of selecting item i:

   P(i) = w_i / Σ_j w_j

   Where:
   - P(i) is the probability of selecting item i
   - w_i is the weight of item i
   - Σ_j w_j is the sum of weights of all j items

2. Regularity Property:

   P(S') ≥ P(S) for any subset S' ⊆ S

   Where:
   - P(S) is the probability of selecting any alternative in set S
   - S' is a subset of S

3. Independence from Irrelevant Alternatives (IIA):

   P(i|S) = P(i|{i,j})

   Where:
   - P(i|S) is the probability of selecting item i given the choice set S
   - P(i|{i,j}) is the probability of selecting item i given only options i and j

### Applications and Significance

Luce's Axiom has important applications in:

1. Economics: It can be used to model consumer preferences for products.

2. Psychology: Particularly in cognitive science, it can model stochastic properties of neural networks.

3. Decision Theory: It provides a framework for understanding how individuals make choices among multiple options.

### Critiques and Limitations

While Luce's Axiom has been influential, it has faced criticism and challenges:

1. Empirical limitations: Experiments have shown that the conditions under which the axiom holds are quite delicate.

2. Violation in certain scenarios: The axiom fails to describe choice behavior when the stimulus set is structured in ways that lead to treating several alternatives as substantially the same.

3. Potential biases: When experimental subjects believe they should employ response alternatives about equally often, it can introduce response bias.

### Conclusion

Despite its limitations, Luce's Axiom remains a significant concept in decision theory and choice modeling. Its simplicity and ease of computation contribute to its continued use, especially as a canonical representation of probabilistic rationality. However, researchers continue to explore alternative theories and models that can better capture real-world choice behaviors in various contexts.
